### 2, 3 e 4) Testes Simples de Pub/Sub
Para realizar os testes descritos no roteiro sem travar a célula do Jupyter, colocaremos o comando de escuta (`mosquitto_sub`) rodando no background. Após enviarmos as mensagens, leremos o arquivo de log gerado e encerraremos o processo de escuta.

```python
%%bash
# Inicia o assinante (Terminal 1) em segundo plano, salvando a saída em um arquivo
nohup mosquitto_sub -t "/teste" -v > teste_simples.log 2>&1 &
sleep 2 # Aguarda o serviço iniciar

# Envia mensagens de teste (Terminal 2) variando parâmetros de host, porta e QoS.
mosquitto_pub -t "/teste" -m "enviando msg"
mosquitto_pub -t "/teste" -h 127.0.0.1 -p 1883 -m "enviado msg2"
mosquitto_pub -t "/teste" -h 127.0.0.1 -p 1883 -m "enviado msg2" -q 2

sleep 2 # Aguarda a recepção
cat teste_simples.log

# Encerra o processo em background
pkill -f mosquitto_sub
```

### 5) MQTT somente para usuários autenticados
Vamos criar o arquivo de senha usando a flag `-b` para automatizar a criação do usuário, e sobrescrever o arquivo de configuração do Mosquitto sem usar editores manuais.

```python
%%bash
SENHA="senha"

# Criar arquivo para armazenar o par <usuário>:<senha> e criar o usuário.
echo $SENHA | sudo -S touch /etc/mosquitto/passwd
echo $SENHA | sudo -S mosquitto_passwd -b /etc/mosquitto/passwd user senha
echo $SENHA | sudo -S chmod 777 /etc/mosquitto/passwd
echo $SENHA | sudo -S chown mosquitto:mosquitto /etc/mosquitto/passwd
```

```python
%%writefile mosquitto.conf
# Editar o arquivo de configuração para só permitir conexões autenticadas.
allow_anonymous false
password_file /etc/mosquitto/passwd
```

```python
%%bash
SENHA="senha"

echo $SENHA | sudo -S mv mosquitto.conf /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S service mosquitto restart
sleep 2

# Teste de autenticação
nohup mosquitto_sub -h localhost -t test -u user -P senha > teste_autenticado.log 2>&1 &
sleep 2
mosquitto_pub -h localhost -t "test" -m "autenticado" -u user -P senha
sleep 2
cat teste_autenticado.log
echo $SENHA | sudo -S pkill -f mosquitto_sub
```

### 6) MQTT com dados criptografados (chave privada e chave pública)
Baixaremos o script gerador, ajustaremos a variável `IPLIST` via linha de comando (`sed`) para não interagir com a tela, e moveremos as chaves para os diretórios corretos.

```python
%%bash
SENHA="senha"

# Instalar o net-tools para que o script utilize o comando ifconfig.
echo $SENHA | sudo -S apt install -y net-tools

# Baixar o script que cria as chaves corretamente.
wget https://raw.githubusercontent.com/owntracks/tools/master/TLS/generate-CA.sh -O generate-CA.sh

# Configurar a variável IPLIST com o endereço do broker MQTT substituindo no arquivo.
sed -i 's/IPLIST=""/IPLIST="127.0.0.1"/g' generate-CA.sh

# Alterar permissões e executar.
echo $SENHA | sudo -S chmod 777 generate-CA.sh
echo $SENHA | sudo -S bash generate-CA.sh

# As chaves geradas para a máquina devem ser movidas para /etc/mosquitto/certs/.
echo $SENHA | sudo -S mv vivobook.crt /etc/mosquitto/certs/mosq-serv.crt
echo $SENHA | sudo -S mv ca.crt /etc/mosquitto/certs/mosq-ca.crt
echo $SENHA | sudo -S mv vivobook.key /etc/mosquitto/certs/mosq-serv.key

# Alterar as permissões e o dono das chaves.
echo $SENHA | sudo -S chmod 777 -R /etc/mosquitto/certs/
echo $SENHA | sudo -S chown mosquitto:mosquitto -R /etc/mosquitto/certs/
```

```python
%%writefile mosquitto_tls.conf
pid_file /var/run/mosquitto/mosquitto.pid
persistence true
persistence_location /var/lib/mosquitto/
log_dest file /var/log/mosquitto/mosquitto.log
allow_anonymous false
password_file /etc/mosquitto/passwd
port 8883
cafile /etc/mosquitto/certs/mosq-ca.crt
certfile /etc/mosquitto/certs/mosq-serv.crt
keyfile /etc/mosquitto/certs/mosq-serv.key
```

```python
%%bash
SENHA="senha"

# Editar /etc/mosquitto/mosquitto.conf para exigir autenticação e criptografia e reiniciar.
echo $SENHA | sudo -S mv mosquitto_tls.conf /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S service mosquitto stop
echo $SENHA | sudo -S service mosquitto start
sleep 2

# Aguardar e enviar mensagens autenticadas e criptografadas.
nohup mosquitto_sub -t "/T1" -v --cafile /etc/mosquitto/certs/mosq-ca.crt -p 8883 -u user -P senha > teste_tls.log 2>&1 &
sleep 2
mosquitto_pub -h 127.0.0.1 -t "/T1" --cafile /etc/mosquitto/certs/mosq-ca.crt -p 8883 -m "autenticado_criptografado" -u user -P senha
sleep 2
cat teste_tls.log
echo $SENHA | sudo -S pkill -f mosquitto_sub
```

### 7) Sensores e atuadores virtuais em Python
Comentar as configurações de segurança, instalar o Python 3 (garantindo pip e build-essential) e a biblioteca paho-mqtt.

```python
%%bash
SENHA="senha"

# Reconfigurar o mosquitto comentando as linhas de autenticação/TLS
echo $SENHA | sudo -S sed -i 's/^allow_anonymous false/#allow_anonymous false/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S sed -i 's/^password_file/#password_file/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S sed -i 's/^port 8883/#port 8883/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S sed -i 's/^cafile/#cafile/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S sed -i 's/^certfile/#certfile/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S sed -i 's/^keyfile/#keyfile/g' /etc/mosquitto/mosquitto.conf
echo $SENHA | sudo -S service mosquitto restart

# Instalar Python 3 e biblioteca paho-mqtt
echo $SENHA | sudo -S apt update
echo $SENHA | sudo -S apt install -y python3 python3-pip build-essential

# Correção: Flag adicionada para permitir a instalação no ambiente global
echo $SENHA | sudo -S pip3 install paho-mqtt --break-system-packages
```

```python
%%writefile sv_6000.py
# 7.1) Sensor virtual publicando temperatura aleatória a cada 5s.
import paho.mqtt.client as mqtt
from random import randint
from time import sleep

AREA_ID = 10
SENSOR_ID = 6000  # usar SENSOR_ID = 5000 para sv_5000.py.

tt = "area/%d/sensor/%s/temperatura" % (AREA_ID, SENSOR_ID)

# Scripts abaixo já usam mqtt.CallbackAPIVersion.VERSION1.
client = mqtt.Client(
    mqtt.CallbackAPIVersion.VERSION1,
    client_id='NODE:%d-%d' % (AREA_ID, SENSOR_ID),
    protocol=mqtt.MQTTv31,
)

client.connect("127.0.0.1", 1883)

while True:
    t = randint(0, 50)
    msg = str(t)
    client.publish(tt, msg, qos=0)
    print(tt + "/" + str(t))
    sleep(5)
```

```python
%%bash
# Executaremos em background para não travar a célula do Jupyter
nohup python3 sv_6000.py > python_sensor.log 2>&1 &
```

```python
%%writefile codigo_arduino/codigo_arduino.ino
// Criar um sketch. Esse código testa MQTT com Python sem segurança.
const int led = 8;
int estadoler = 10;

void setup() {
  Serial.begin(9600);
  pinMode(led, OUTPUT); 
  while (!Serial) {
    ;
  }
}

String leStringSerial() {
  String conteudo = "";
  char caractere;
  while (Serial.available() > 0) { 
    caractere = Serial.read(); 
    if (caractere != '\n') { 
      conteudo.concat(caractere); 
    }
    delay(10); 
  }
  Serial.print("Recebi: ");
  Serial.println(conteudo);
  return conteudo;
}

void loop() {
  while (Serial.available() > 0) { 
    int est = digitalRead(led);
    String recebido = leStringSerial(); 
    if (recebido == "L1") { 
      if (est == 0) {
        digitalWrite(led, HIGH); 
      } else if (est == 1) {
        digitalWrite(led, LOW); 
      }
    }
    delay(1); 
  }
}
```
